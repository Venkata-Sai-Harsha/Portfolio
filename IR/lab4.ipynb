```python
import numpy as np
from collections import defaultdict
import math

class NaiveBayesClassifier:
    def __init__(self):
        self.classes = None
        self.vocab = set()
        self.class_word_counts = defaultdict(lambda: defaultdict(int))
        self.class_doc_counts = defaultdict(int)
        self.total_docs = 0

    def train(self, documents, labels):
        self.classes = set(labels)
        self.total_docs = len(labels)

        for doc, label in zip(documents, labels):
            self.class_doc_counts[label] += 1
            for word in doc:
                self.vocab.add(word)
                self.class_word_counts[label][word] += 1

    def predict(self, document):

        scores = {}
        vocab_size = len(self.vocab)

        for cls in self.classes:
            log_prob = math.log(self.class_doc_counts[cls] / self.total_docs)

            total_words_in_class = sum(self.class_word_counts[cls].values())
            for word in document:
                word_freq = self.class_word_counts[cls][word]
                log_prob += math.log((word_freq + 1) / (total_words_in_class + vocab_size))

            scores[cls] = log_prob

        return max(scores, key=scores.get)

if __name__ == "__main__":
    documents = [
        ["love", "this", "movie"],
        ["hate", "this", "movie"],
        ["amazing", "cinematography"],
        ["boring", "and", "dull"],
        ["great", "storyline"],
        ["terrible", "acting"]
    ]
    labels = ["positive", "negative", "positive", "negative", "positive", "negative"]

    classifier = NaiveBayesClassifier()
    classifier.train(documents, labels)

    test_doc = ["great", "cinematography"]
    predicted_label = classifier.predict(test_doc)
    print(f"The predicted label for document '{test_doc}' is '{predicted_label}'")
```
