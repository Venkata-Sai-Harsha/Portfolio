```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Sample corpus
documents = [
    "Machine learning is fascinating.",
    "Deep learning is a subset of machine learning.",
    "Artificial intelligence drives innovation.",
    "Data science overlaps with artificial intelligence.",
    "Mathematics is fundamental for machine learning.",
    "Cluster algorithms like K-means are unsupervised."
]

# Step 1: Preprocess and vectorize
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(documents)

# Step 2: Apply K-Means
num_clusters = 2
kmeans = KMeans(n_clusters=num_clusters, random_state=42)
kmeans.fit(X)

# Assign clusters to documents
clusters = kmeans.labels_

# Step 3: Dimensionality reduction for visualization
pca = PCA(n_components=2)
reduced_data = pca.fit_transform(X.toarray())

# Visualize clusters
plt.scatter(reduced_data[:, 0], reduced_data[:, 1], c=clusters, cmap='viridis', marker='o')
for i, doc in enumerate(documents):
    plt.text(reduced_data[i, 0], reduced_data[i, 1], f'Doc {i+1}')
plt.title("K-Means Document Clustering")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.colorbar(label='Cluster')
plt.show()
```
